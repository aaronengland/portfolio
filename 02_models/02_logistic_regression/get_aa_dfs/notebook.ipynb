#### Indirect

```python
str_model = '09_60_in_720'

# get model features
list_cols = [
    'feature',
]
str_filename = 'df_feat_imp.csv'
str_local_path = f'../{str_model}/02_model/output/{str_filename}'
df = pd.read_csv(str_local_path, usecols=list_cols)
df['feature'] = df['feature'].apply(
    lambda x: x.split('_binned')[0],
)

# get the AA dict
str_filename = 'df_aa.csv'
str_local_path = f'../{str_model}/06_parser/input/{str_filename}'
df_tmp = pd.read_csv(str_local_path)

# join
df = pd.merge(
    left=df,
    right=df_tmp,
    on='feature',
    how='left',
)

# replace
df['model'] = 'gen-xiii'

# drop
df.drop('Change Reason/Comment', axis=1, inplace=True)

# save
str_filename = 'df_aa_indirect.csv'
str_local_path = f'{str_dirname_output}/{str_filename}'
df.to_csv(str_local_path, index=False)

# show
df
```

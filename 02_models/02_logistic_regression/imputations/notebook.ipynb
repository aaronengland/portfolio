```python
import os
import pandas as pd
import numpy as np
import datetime as dt
import pickle
```

```python
print(f'Last run date: {dt.datetime.now()}')
```

#### Constants

```python
str_dirname_output = './output'
```

#### Make output directory

```python
try:
    os.mkdir(str_dirname_output)
except:
    pass
```

#### Indirect

```python
# get feature importance
str_filename = 'df_feat_imp.csv'
str_local_path = f'../09_60_in_720/02_model/output/{str_filename}'
df = pd.read_csv(str_local_path)

# get imputation dictionary
str_filename = 'dict_impute.pkl'
str_local_path = f'../10_60_in_720_test/02_model/output/{str_filename}'
dict_impute = pickle.load(open(str_local_path, 'rb'))

# add suffix
dict_impute = {f'{key}_binned': val for key, val in dict_impute.items()}

# map
df['imputation'] = df['feature'].map(dict_impute)

# rm suffix
df['feature'] = df['feature'].apply(
    lambda x: x.split('_binned')[0],
)

# reorder
list_cols = [
    'feature',
    'importance',
    'importance_abs',
    'imputation',
    'description',
]
df = df[list_cols].copy()

# save
str_filename = 'df_imputations_indirect.csv'
str_local_path = f'{str_dirname_output}/{str_filename}'
df.to_csv(str_local_path, index=False)

# show
df
```

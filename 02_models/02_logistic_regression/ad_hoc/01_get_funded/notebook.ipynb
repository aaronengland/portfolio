```python
import os
import pandas as pd
import numpy as np
```

```python
str_dirname_output = './output'
```

```python
try:
    os.mkdir(str_dirname_output)
except:
    pass
```

```python
# get the funded
str_filename = 'df_all_funded.csv'
str_local_path = f'./input/{str_filename}'
df = pd.read_csv(str_local_path)
df['Account ID'] = df['Account ID'].astype(int)
list_accounts = list(df['Account ID'])
```

```python
# get the swap in swap out
str_filename = 'df_swap_in_swap_out.csv'
str_local_path = f'../../09_60_in_720/10_swap_in_swap_out/output/{str_filename}'
df_tmp = pd.read_csv(str_local_path)
df_tmp['accountid'] = df_tmp['accountid'].astype(int)
df_tmp['funded'] = df_tmp['accountid'].apply(
    lambda x: 1 if x in list_accounts else 0,
)
df_tmp = df_tmp[df_tmp['funded'] == 1].copy()
df_tmp
```

```python
# save
str_filename = 'df_w_funded_tag.csv'
str_local_path = f'{str_dirname_output}/{str_filename}'
df_tmp.to_csv(str_local_path, index=False)
```

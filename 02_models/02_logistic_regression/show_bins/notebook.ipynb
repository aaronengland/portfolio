#### Indirect

```python
# get features in model
str_filename = 'df_feat_imp.csv'
str_local_path = f'../09_60_in_720/02_model/output/{str_filename}'
df_feat_imp = pd.read_csv(str_local_path)
list_cols_in_model = list(df_feat_imp['feature'])
list_cols_in_model = [f'{col.split("_binned")[0]}' for col in list_cols_in_model]

# get bins
str_filename = 'dict_bins.pkl'
str_local_path = f'../10_60_in_720_test/02_model/output/{str_filename}'
dict_bins = pickle.load(open(str_local_path, 'rb'))

# show binning table
list_df = []
for key, val in tqdm(dict_bins.items()):
    if key in list_cols_in_model:
        df_bins = val.binning_table.build()
        list_df.append(df_bins)
        df_bins['Feature'] = key
    else:
        pass

# concat
df = pd.concat(list_df)

# get the aa dict
list_cols = [
    'feature',
    'reason',
    'description',
]
str_filename = 'df_aa.csv'
str_local_path = f'../09_60_in_720/06_parser/input/{str_filename}'
df_aa = pd.read_csv(str_local_path, usecols=list_cols)

# rename
dict_rename = {
    'feature': 'Feature',
    'reason': 'Reason',
    'description': 'Description',
}
df_aa.rename(columns=dict_rename, inplace=True)

# join
df = pd.merge(
    left=df,
    right=df_aa,
    on='Feature',
    how='left',
)

# reorder
list_cols = [
    'Feature',
    'Bin',
    'Count',
    'Count (%)',
    'Non-event',
    'Event',
    'Event rate',
    'WoE',
    'IV',
    'JS',
    'Reason',
    'Description',
]
df = df[list_cols].copy()

# save
str_filename = 'df_bins_indirect.csv'
str_local_path = f'{str_dirname_output}/{str_filename}'
df.to_csv(str_local_path, index=False)

# show
df
```

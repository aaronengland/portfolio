```python
import os
import boto3
import shutil
import pandas as pd
```

### Functions

```python
# download from s3
def download_from_s3(str_local_path, str_bucket_path, str_project):
    boto3.client('s3').download_file(str_project, str_bucket_path, str_local_path)
```

```python
# upload to s3
def upload_to_s3(str_local_path, str_bucket_path, str_project):
    boto3.resource('s3').Bucket(str_project).Object(str_bucket_path).upload_file(str_local_path)
```

```python
# get table
def get_table(str_feature):
    if '__app' in str_feature:
        return 'Application'
    elif '__income' in str_feature:
        return 'Income'
    elif '__ln' in str_feature:
        return 'LexisNexis'
    elif '__tu' in str_feature:
        return 'TransUnion'
    elif 'ENG-' in str_feature:
        return 'Multiple'
    else:
        return 'Unknown'
```

### Descriptions

```python
str_filename = 'df_descriptions.csv'
str_local_path = f'../../ad_hoc/06_get_descriptions_all/output/{str_variant}/{str_filename}'
# read
df = pd.read_csv(str_local_path)
# rename
dict_rename = {
    'feature': 'Feature Name',
    'description': 'Description',
}
df.rename(columns=dict_rename, inplace=True)
# get table
df['Table'] = df['Feature Name'].apply(get_table)
# subset
list_cols = [
    'Feature Name',
    'Description',
    'Table',
]
df = df[list_cols]
# write to local app
str_destination = f'../02_local_app/{str_filename}'
df.to_csv(str_destination, index=False)

# show
df
```

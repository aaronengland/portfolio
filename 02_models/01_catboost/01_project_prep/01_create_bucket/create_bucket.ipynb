# Create Bucket

```python
import os
import boto3
```

## Constants

```python
str_project = os.getcwd().split('/')[-3].replace('_','-')
print(f'Project: {str_project}')
```

## Get region

```python
# region
str_region = boto3.client('s3').meta.region_name
print(f'Region: {str_region}')
```

## Create bucket

```python
# initialize class
cls_client = boto3.client('s3')

# config dictionary
dict_config = {
    'LocationConstraint': str_region,
}

# create bucket
cls_client.create_bucket(
    ACL='private',
    Bucket=str_project,
    CreateBucketConfiguration=dict_config,
)
```

```python
import os
import pandas as pd
import functions as func
import numpy as np
from tqdm import tqdm
```

### Constants

```python
str_project = os.getcwd().split('/')[4].replace('_','-')
print(f'Project: {str_project}')
str_dirname_output = './output'
```

### Output

```python
try:
    os.mkdir(str_dirname_output)
except:
    pass
```

### Empty list

```python
%%time

list_cols_all = []
for str_df in tqdm (['train','valid','test']):
    # print
    print(f'Data set: {str_df}')
    print('')
    
    # import
    str_filename = f'df_{str_df}_raw.gzip'
    str_uri = f's3://{str_project}/03_pricing_lgd/01_data_prep/03_train_valid_test_split/{str_filename}'
    # read from s3
    df = pd.read_parquet(str_uri)
    # replace
    df.replace(['NaN','nan'], np.nan, inplace=True)
    
    # check all nan
    list_cols = func.find_all_nan_columns(
        df=df,
    )
    list_cols_all.extend(list_cols)
    print(f'There are {len(list_cols)} features in the {str_df} df that are 100% NaN')
    for a, col in enumerate(list_cols):
        print(f'{a+1} - {col}')
    print('')
    
    # check no var
    list_cols = func.find_no_variance_columns(
        df=df,
    )
    list_cols_all.extend(list_cols)
    print(f'There are {len(list_cols)} features in the {str_df} df that have no variance')
    for a, col in enumerate(list_cols):
        print(f'{a+1} - {col}')
    print('')
    
    # check redundancy
    list_cols = func.find_redundant_columns(
        df=df,
    )
    list_cols_all.extend(list_cols)
    print(f'There are {len(list_cols)} features in the {str_df} df that are redundant')
    for a, col in enumerate(list_cols):
        print(f'{a+1} - {col}')
    print('')
```

### Remove duplicates

```python
print(f'Before removing duplicates, there were {len(list_cols_all)} features to remove')
print('')

# rm dups
list_cols_all = list(dict.fromkeys(list_cols_all))
print(f'After removing duplicates, there are {len(list_cols_all)} features to remove')
for a, col in enumerate(list_cols_all):
    print(f'{a+1} - {col}')
```

### Save

```python
%%time

df = pd.DataFrame({'feature': list_cols_all})
str_filename = 'df_cols_drop.csv'
str_local_path = f'{str_dirname_output}/{str_filename}'
df.to_csv(str_local_path, index=False)
```

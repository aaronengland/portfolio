```python
import os
import pandas as pd
import numpy as np
```

### Constants

```python
str_dirname_output = './output'
```

### Output directory

```python
try:
    os.mkdir(str_dirname_output)
except:
    pass
```

### Get training means

```python
str_filename = 'df_mean.csv'
str_local_path = f'../03_training_data_frequency/output/{str_filename}'
df = pd.read_csv(str_local_path)
# rename
dict_rename = {
    'mean': 'mean_train',
}
df.rename(columns=dict_rename, inplace=True)
df
```

### Get test means

```python
str_filename = 'df_mean.csv'
str_local_path = f'../02_test_data_frequency/output/{str_filename}'
df_tmp = pd.read_csv(str_local_path)
# rename
dict_rename = {
    'mean': 'mean_test',
}
df_tmp.rename(columns=dict_rename, inplace=True)
df_tmp
```

### Join

```python
df = pd.merge(
    left=df,
    right=df_tmp,
    on='flag',
    how='left',
)
df
```

### Save

```python
str_filename = 'df_mean.csv'
str_local_path = f'{str_dirname_output}/{str_filename}'
df.to_csv(str_local_path, index=False)
```

```python
import boto3
from pprint import pprint
from tqdm import tqdm
import time
```

### Constants

```python
str_substring = 'genxii-analysis-analysis-db'

# never delete these
list_str_ignore = [
#     'queue-genxii-pull-payloads-1',
#     'queue-genxii-parse-payloads-1',
#     'env-genxi-pull-payloads-1',
#     'env-genxii-pull-payloads-1',
]
```

### Initialize batch client

```python
# init client
cls_client = boto3.client('batch')
```

### Disable and delete queues

```python
# get the job queues
list_dict_queues = cls_client.describe_job_queues()['jobQueues']
list_str_queues = [dict_queue['jobQueueName'] for dict_queue in list_dict_queues]
list_str_queues = [str_queue for str_queue in list_str_queues if str_substring in str_queue]
list_str_queues = [str_queue for str_queue in list_str_queues if str_queue not in list_str_ignore]
list_str_queues
```

```python
# disable
for str_queue in tqdm(list_str_queues):
    dict_response = cls_client.update_job_queue(
        jobQueue=str_queue,
        state='DISABLED',
    )
```

```python
# delete
for str_queue in tqdm(list_str_queues):
    while True:
        try:
            dict_response = cls_client.delete_job_queue(
                jobQueue=str_queue,
            )
            break
        except:
            time.sleep(1)
```

### Disable and delete batch compute environments

```python
# List the existing compute environments
list_dict_compute_envs = cls_client.describe_compute_environments()['computeEnvironments']
list_str_envs = [dict_env['computeEnvironmentName'] for dict_env in list_dict_compute_envs]
list_str_envs = [str_env for str_env in list_str_envs if str_substring in str_env]
list_str_envs = [str_env for str_env in list_str_envs if str_env not in list_str_ignore]
pprint(list_str_envs)
```

```python
# disable
for str_env in tqdm(list_str_envs):
    dict_response = cls_client.update_compute_environment(
        computeEnvironment=str_env,
        state='DISABLED',
    )
```

```python
# delete
for str_env in tqdm(list_str_envs):
    while True:
        try:
            dict_response = cls_client.delete_compute_environment(
                computeEnvironment=str_env,
            )
            break
        except:
            time.sleep(1)
```

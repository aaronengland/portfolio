```python
import os
import pandas as pd
import numpy as np
```

### Functions

```python
# get percent difference
def get_percent_difference(val1, val2):
    flt_diff = abs(val1 - val2)
    flt_mean = np.mean([val1, val2])
    prop_diff = flt_diff / flt_mean
    return prop_diff * 100
```

### Constants

```python
str_dirname_output = './output'
str_variant = 'model6'
```

### Output directory

```python
try:
    os.mkdir(str_dirname_output)
except:
    pass
```

### Variant directory

```python
try:
    os.mkdir(f'{str_dirname_output}/{str_variant}')
except:
    pass
```

### Get descriptives from PD test data

```python
str_filename = 'df_descriptives.csv'
str_local_path = f'../get_predictions/03_eda/output/{str_variant}/{str_filename}'
df_pd = pd.read_csv(str_local_path)

# add suffix
df_pd.columns = [f'{col}__pd' for col in df_pd.columns]

# rename
dict_rename = {
    'feature__pd': 'feature',
}
df_pd.rename(columns=dict_rename, inplace=True)

# show
df_pd
```

### Get descriptives from parsed data

```python
str_filename = 'df_descriptives.csv'
str_local_path = f'../../08_retro_scoring/07_analysis/02_eda/output/{str_variant}/{str_filename}'
df_parsed = pd.read_csv(str_local_path)

# add suffix
df_parsed.columns = [f'{col}__parse' for col in df_parsed.columns]

# rename
dict_rename = {
    'feature__parse': 'feature',
}
df_parsed.rename(columns=dict_rename, inplace=True)

# show
df_parsed
```

### Join

```python
df = pd.merge(
    left=df_pd,
    right=df_parsed,
    on='feature',
    how='inner',
)

# show
df
```

### Check means

```python
%%time

df['mean_diff'] = df.apply(
    lambda x: get_percent_difference(
        val1=x['mean__pd'], 
        val2=x['mean__parse'],
    ),
    axis=1,
)

# make absolute value
df['mean_diff'] = df['mean_diff'].abs()

# sort
df.sort_values(by='mean_diff', ascending=False, inplace=True)

# list cols
list_cols = [
    'feature',
    'mean__pd',
    'mean__parse',
    'mean_diff',
]

# subset
df_tmp = df[list_cols].copy()

# save
str_filename = 'df_mean_diff.csv'
str_local_path = f'{str_dirname_output}/{str_variant}/{str_filename}'
df_tmp.to_csv(str_local_path)

# show
df_tmp
```

### Get medians

```python
%%time

df['median_diff'] = df.apply(
    lambda x: get_percent_difference(
        val1=x['median__pd'], 
        val2=x['median__parse'],
    ),
    axis=1,
)

# make absolute value
df['median_diff'] = df['median_diff'].abs()

# sort
df.sort_values(by='median_diff', ascending=False, inplace=True)

# list cols
list_cols = [
    'feature',
    'median__pd',
    'median__parse',
    'median_diff',
]

# subset
df_tmp = df[list_cols].copy()

# save
str_filename = 'df_median_diff.csv'
str_local_path = f'{str_dirname_output}/{str_variant}/{str_filename}'
df_tmp.to_csv(str_local_path)

# show
df_tmp
```

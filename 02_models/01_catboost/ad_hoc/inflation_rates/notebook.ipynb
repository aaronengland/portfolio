### Get rates

```python
int_starting_value = 100
list_int_years = [
    2013,
    2014,
    2015,
    2016,
    2017,
    2018,
    2019,
    2020,
    2021,
    2022,
]
dict_rates = {}
for int_year in list_int_years:
    # get new value
    flt_new_value = cpi.inflate(int_starting_value, int_year, to=2022)
    # get rate
    flt_rate = flt_new_value / int_starting_value
    # assign
    dict_rates[int_year] = flt_rate

# show
dict_rates
```

### Get factors for future years assuming a 3% increase annually

```python
flt_factor_start = 1.0
flt_inflation_rate = 0.03 # 3%
list_int_years = [
    2023,
    2024,
    2025,
    2026,
    2027,
]
dict_factors = {}
for int_year in list_int_years:
    # get factor
    flt_factor = flt_factor_start * (1 - flt_inflation_rate)
    # assign
    dict_factors[int_year] = flt_factor
    # update
    flt_factor_start = flt_factor
# show
dict_factors
```

### Combine

```python
dict_inflation_rates = {**dict_rates, **dict_factors}

# show
dict_inflation_rates
```

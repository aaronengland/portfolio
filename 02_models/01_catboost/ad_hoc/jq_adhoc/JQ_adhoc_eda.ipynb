```python
list_cols_check = [
'ENG-payment_to_income'
,'payment__app'
,'fltgrossmonthly__income_sum'
,'amtfinanced__app'
,'bookvalue__app'
,'ENG-loan_to_value'
,'ENG-vehicle_age'
,'vehicleyear__app'
,'ENG-dealership_age'
#,'dealerstampcreation__app'
,'intterm__app'
#,'ENG-applicationdate__app_month'
#,'ENG-applicationdate__app_quarter'
]
```

```python
def check_no_variance(df, columns):
    result = {}
    for col in columns:
        variance = df[col].var()
        has_variance = variance != 0
        result[col] = not has_variance
    return result
```

```python
result = check_no_variance(df, list_cols_check)
print(result)
```

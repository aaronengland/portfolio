```python
str_uri = 's3://20231010-gen-xii/ad_hoc/09_policy_simulation/df.gzip'
df = pd.read_parquet(str_uri)
df
```

#### Rm accounts with app date before 02-26-2025

```python
str_dtm_effective_date = '2/26/2025 00:00:00'
dtm_effective_date = dt.datetime.strptime(str_dtm_effective_date, '%m/%d/%Y %H:%M:%S')

df['applicationdate__app'] = pd.to_datetime(df['applicationdate__app'])
df = df[df['applicationdate__app'] >= dtm_effective_date].copy()
df
```

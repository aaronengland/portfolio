## Imports

```python
import os
import pandas as pd
import numpy as np
```

## Constants

```python
str_project = os.getcwd().split('\\')[5]
print(f'Project: {str_project}')
str_task = os.getcwd().split('\\')[6]
print(f'Task: {str_task}')
str_subtask = os.getcwd().split('\\')[7]
print(f'Subtask: {str_subtask}')

str_dirname_output = './output'
```

## Output Dir

```python
#make an output for the train, valid, test files
try:
    os.mkdir(str_dirname_output)
except FileExistsError:
    pass
```

## Read train.csv

```python
str_filename = 'train.csv'
str_local_path = f'./input/{str_filename}'
df = pd.read_csv(str_local_path)
df.head()
```

## Sort the Df

```python
#sort by id, ascending such that model is trained on the first 60% of accounts
df_tmp = df.sort_values(by=['SK_ID_CURR'], ascending=True)
#insert a row number column such that we can split the df into 60-20-20 based on total rows
df_tmp.insert(0, 'RowNum', np.arange(df_tmp.shape[0]))
#show
df_tmp.head()
```

```python
df_tmp.shape
```

## Split Df into Train, Valid, Test

### Training

```python
df_training = df_tmp.query('0 <= RowNum < 75000')
df_training.tail()
```

```python
df_training.shape
```

### Validation

```python
df_validation = df_tmp.query('75000 <= RowNum < 100000')
df_validation.tail()
```

```python
df_validation.shape
```

### Testing

```python
df_testing = df_tmp.query('RowNum >= 100000')
df_testing.tail()
```

```python
df_testing.shape
```

## Output Df's as CSV

### Training

```python
str_filename = 'df_training.csv'
str_local_path = f'{str_dirname_output}/{str_filename}'
df_training.to_csv(str_local_path, index=False)
```

### Validation

```python
str_filename = 'df_validation.csv'
str_local_path = f'{str_dirname_output}/{str_filename}'
df_validation.to_csv(str_local_path, index=False)
```

### Testing

```python
str_filename = 'df_testing.csv'
str_local_path = f'{str_dirname_output}/{str_filename}'
df_testing.to_csv(str_local_path, index=False)
```

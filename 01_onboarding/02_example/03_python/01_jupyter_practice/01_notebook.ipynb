```python
import pandas as pd
import pyodbc
import os
```

#### read sql in code

```python
# Use """ to designate SQL query as a string
str_query = """
SELECT 
bigAccountId,
FundingMonth,
LoanStatus,
strTier,
DebtorState
FROM electra.riskdb.analytics.tbltempstaticpool
WHERE LoanStatus = 'C' AND DebtorState IN ('CA', 'AZ')
ORDER BY FundingMonth desc
"""
```

```python
# establish SQL connection
conn = pyodbc.connect(
    'Driver={SQL Server};'
    'Server=electra;'
    'Database=pfsdb;'
    'Trusted_Connection=yes;'
)

data = pd.read_sql_query(
    sql=str_query,
    con=conn,
)


# close the connection 
conn.close()

# how variables work
# conn.close() == pyodbc.close()
```

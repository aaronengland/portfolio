```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import pyodbc
import numpy as np
import pickle
```

## Functions

```python
class GetData:
    def __init__(self):
        pass
    def read_sql_script(self, str_local_path):
        str_query = open(str_local_path, 'r').read()
        # save to object
        self.str_query = str_query
        # return object
        return self
    def pull_data_from_sql(self, conn, bool_close_conn=True):
        df = pd.read_sql_query(
            sql=self.str_query,
            con=conn,
        )
        if bool_close_conn:
            conn.close()
        else:
            pass
        return df
```

## DimScoreCard Gen11 Query

```python
# initialize class and gather data from the dimscorecard table
cls_get_data = GetData()

#read query
str_filename = 'DimScoreCardGen11.sql'
str_local_path = f'./sql/{str_filename}'
cls_get_data.read_sql_script(
    str_local_path=str_local_path,
)
# pull data
conn = pyodbc.connect(
    'Driver={SQL Server};'
    'Server=electra;'
    'Database=pfsdb;'
    'Trusted_connection=yes;'
)
dfgen11 = cls_get_data.pull_data_from_sql(
    conn=conn,
    bool_close_conn=True,
)
dfgen11
```

```python
#drop row num values
dfgen11sort.drop(columns=['rownum'])
```

## DimScoreCard Gen12 Query

```python
# initialize class and gather data from the dimscorecard table
cls_get_data = GetData()

#read query
str_filename = 'DimScoreCardGen12.sql'
str_local_path = f'./sql/{str_filename}'
cls_get_data.read_sql_script(
    str_local_path=str_local_path,
)
# pull data
conn = pyodbc.connect(
    'Driver={SQL Server};'
    'Server=electra;'
    'Database=pfsdb;'
    'Trusted_connection=yes;'
)
dfgen12 = cls_get_data.pull_data_from_sql(
    conn=conn,
    bool_close_conn=True,
)
dfgen12
```

```python
# sort dataframe by accountkey and dtmstampcreation
dfgen12 = dfgen12.sort_values(by=['intAccountKey','dtmStampCreation'], ascending=[True,False])
```

```python
# apply row numbers to all of the most recent values based on account
dfgen12['rownum'] = dfgen12.groupby('intAccountKey').cumcount() + 1
```

## Join Gen11 and Gen12 Tables

```python
# merge gen tables on account key
dfgentable = pd.merge(dfgen11sort, dfgen12sort, left_on='intAccountKey', right_on=['intAccountKey'], how='inner')
dfgentable
```

## Join GenTable and AccountTable

```python
# merge gen table and tempstaticpool table
dffullmerge = pd.merge(dftemp, dfgentable, left_on='bigAccountId', right_on='intAccountKey', how='left')
dffullmerge
```

```python
# drop unneeded columns
dffullmerge.drop(columns='intAccountKey', inplace=True)
dffullmerge.drop(columns='dtmStampCreation_x', inplace=True)
dffullmerge.drop(columns='strScoreCardVersion_x', inplace=True)
dffullmerge.drop(columns='rownum_x', inplace=True)
dffullmerge.drop(columns='dtmStampCreation_y', inplace=True)
dffullmerge.drop(columns='strScoreCardVersion_y', inplace=True)
dffullmerge.drop(columns='rownum_y', inplace=True)
dffullmerge
```

```python
# rename Debtor Scores columns to gen 11 and 12
dffullmerge.rename(columns={'fltDebtorScore_x' : 'Gen 11'}, inplace=True)
dffullmerge.rename(columns={'fltDebtorScore_y' : 'Gen 12'}, inplace=True)
```

```python
# creates extra column to pick gen version used
dffullmerge['Gen'] = np.where(((dffullmerge['bigAccountId'] % 10 == 0) | (dffullmerge['bigAccountId'] % 10 == 5)) & (dffullmerge['dtmStampCreation'] >= '2024-04-15'), 'Gen12', 'Gen11')
dffullmerge
```

```python
# reset index and sort for both df_check and dffullmerge for future comparison
df_check = df_check.sort_values(by=['bigAccountId','dtmStampCreation'], ascending=[True,False])
df_check = df_check.reset_index(drop=True)
dffullmerge = dffullmerge.sort_values(by=['bigAccountId','dtmStampCreation'],ascending=[True,False])
dffullmerge = dffullmerge.reset_index(drop=True)
```

```python
# searches both dataframes and looks for differences
count = 0
for row in range(df_check.shape[0]):
    for col in df_check.columns:
        val1 = df_check.at[row,col]
        val2 = dffullmerge.at[row,col]
        if val1 != val2 and not (pd.isna(val1) and pd.isna(val2)):
            print(f"Difference found at row {row}, column {col}: df_check has {df_check.at[row,col]}, dffullmerge has {dffullmerge.at[row,col]}")
            count += 1          
```

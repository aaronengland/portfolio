```python
names = ['bob', 'peter', 'patrick','hank']

guesscount = 0
while True:
    guess1 = input('Who was the famous builder on TV? (bob, peter, patrick, hank) |')
    guess1 == guess1.lower()
    if guess1 != names[0]:
        print('Are you sure about that?')
        guesscount += 1
    else:
        print('You got It')
        guesscount += 1
        break
while True:
    guess2 = input('Who the king of the hill? (bob, peter, patrick, hank) |')
    guess2 == guess2.lower()
    if guess2 != names[3]:
        print('Are you sure about that?')
        guesscount += 1
    else:
        print('You got It')
        guesscount += 1
        break
while True:
    guess3 = input('Who is the most family guy? (bob, peter, patrick, hank) |')
    guess3 == guess3.lower()
    if guess3 != names[1]:
        print('Are you sure about that?')
        guesscount += 1
    else:
        print('You got It')
        guesscount += 1
        break
while True:
    guess4 = input('who lives under a rock? (bob, peter, patrick, hank) |')
    guess4 == guess4.lower()
    if guess4 != names[2]:
        print('Are you sure about that?')
        guesscount += 1
    else:
        print('You got It')
        guesscount += 1
        break
score = 4 / guesscount * 100
print('You completed the quiz! Your grade was: ',score,'%')                
```

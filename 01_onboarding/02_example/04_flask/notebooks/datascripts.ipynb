## Functions

```python
class GetData:
    def __init__(self):
        pass
    def read_sql_script(self, str_local_path):
        str_query = open(str_local_path, 'r').read()
        # save to object
        self.str_query = str_query
        # return object
        return self
    def pull_data_from_sql(self, conn, bool_close_conn=True):
        df = pd.read_sql_query(
            sql=self.str_query,
            con=conn,
        )
        if bool_close_conn:
            conn.close()
        else:
            pass
        return df
```

## Pull  DEALER Data

```python
# initialize class 
cls_get_data = GetData()

#read query
str_filename = 'dealerquery.sql'
str_local_path = f'./sql/{str_filename}'
cls_get_data.read_sql_script(
    str_local_path=str_local_path,
)
# pull data
conn = pyodbc.connect(
    'Driver={SQL Server};'
    'Server=electra;'
    'Database=pfsdb;'
    'Trusted_connection=yes;'
)
df = cls_get_data.pull_data_from_sql(
    conn=conn,
    bool_close_conn=True,
)
df
```

```python
# pull applications amounts from df
dfdealerapps = df[['strName','YTDApplicationcount']]
print(dfdealerapps)
# save output
str_filename = 'dealerapps.gzip'
dfdealerapps.to_parquet(f'./output/{str_filename}')
# save memory
del dfdealerapps
```

```python
# pull approved amounts from df
dfdealerapproved = df[['strName','YTDApprovedCount']]
print(dfdealerapproved)
# save output
str_filename = 'dealerapproved.gzip'
dfdealerapproved.to_parquet(f'./output/{str_filename}')
# save memory
del dfdealerapproved
```

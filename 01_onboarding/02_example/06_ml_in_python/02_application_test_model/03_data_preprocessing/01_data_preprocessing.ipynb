```python
# examine text columns for outliers
text_columns = df_train.select_dtypes(include=['category','object'])

count_results = {}
# go through columns getting value countes for each value
for column in text_columns.columns:
    value_counts = text_columns[column].value_counts()
    count_results[column] = value_counts
    
# prints out value counts for each column for each value
for column, counts in count_results.items():
    print(f"Counts for coulmn: {column}")
    for value, count in counts.items():
        print(f"{value}: {count}")
    print("\n" + "--"*40 + "\n")
```

```python
# changes outlier GENDER code to Majority 
df_train.loc[df_train['CODE_GENDER'] == 'XNA', 'CODE_GENDER'] = 'F'
```

```python
# search the dataframes for columns that are missing values and show it as a %
missing = df_train.isnull().mean() * 100
print(missing[missing > 0])
# removes columns that are missing over 65 % of values (changes to 75%)
cols_to_remove = missing[missing > 75].index
print(cols_to_remove)
# drop columns 
df_train_clean = df_train.drop(columns=cols_to_remove)
df_train_clean.shape
```

```python
# Duplicate column drop on test set
df_test_clean = df_test.drop(columns=cols_to_remove)
# show shape of dataset (has 1 less columns due to missing target col)
df_test_clean.shape

#df_valid_clean = df_valid.drop(columns=cols_to_remove)
#df_valid_clean.shape
```

```python
# Search Numerical columns for outliers in training set
def find_outliers(df_train_clean, threshold=4):
    outliers = {}
    
    for col in df_train_clean.select_dtypes(include=[np.number]).columns:
        mean = df_train_clean[col].mean()
        std = df_train_clean[col].std()
        upper_bound = mean + threshold * std
        lower_bound = mean - threshold * std
        
        outliers[col] = df_train_clean[(df_train_clean[col] > upper_bound) | (df_train_clean[col] < lower_bound)][col]
    
    return outliers

outliers = find_outliers(df_train_clean, threshold=4)

for col, outlier_values in outliers.items():
    print(f"Outliers in columns {col}:")
    print(outlier_values)
```

```python
# adjust for outliers within the Numerical columns
df_train.loc[df_train['AMT_INCOME_TOTAL'] == 117000000, 'AMT_INCOME_TOTAL'] = 1170000
```

```python
# setting numerical columns and categorical columns to variables
numcolstrain = df_train_clean.select_dtypes(include=['float64','int64']).columns
catcolstrain = df_train_clean.select_dtypes(include=['object']).columns

numcolstest = df_test_clean.select_dtypes(include=['float64','int64']).columns
catcolstest = df_test_clean.select_dtypes(include=['object']).columns

# filling missing values with 0 and unknown (future imputing done at a later time MODE was best results)
#df_train_clean[numcolstrain] = df_train_clean[numcolstrain].fillna(df_train_clean[numcolstrain].mean())
df_train_clean[numcolstrain] = df_train_clean[numcolstrain].fillna(0)
df_train_clean[catcolstrain] = df_train_clean[catcolstrain].fillna('Unknown')

#df_test_clean[numcolstest] = df_test_clean[numcolstest].fillna(df_test_clean[numcolstest].mean())
df_test_clean[numcolstest] = df_test_clean[numcolstest].fillna(0)
df_test_clean[catcolstest] = df_test_clean[catcolstest].fillna('Unknown')
```

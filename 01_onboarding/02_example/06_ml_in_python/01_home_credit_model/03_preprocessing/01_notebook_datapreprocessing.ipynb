# Clean Data Before Splitting

```python
missing = df.isnull().mean() * 100
print(missing[missing > 0])
cols_to_remove = missing[missing > 65].index
dfclean = df.drop(columns=cols_to_remove)
numcols = dfclean.select_dtypes(include=['float64','int64']).columns
catcols = dfclean.select_dtypes(include=['object']).columns
dfclean[numcols] = dfclean[numcols].fillna(0)
dfclean[catcols] = dfclean[catcols].fillna('Unknown')
```

```python
# Locate Text columns and search for outliers
text_columns = dfclean.select_dtypes(include=['category', 'object'])

count_results = {}

for column in text_columns.columns:
    value_counts = text_columns[column].value_counts()
    count_results[column] = value_counts

for column, counts in count_results.items():
    print(f"Counts for Column: {column}")
    for value, count in counts.items():  
        print(f"{value}: {count}")
    print("\n" + "--"*40 + "\n")
    
```

```python
dfbur = pd.read_csv('./input/bureau.csv')
dfbur
```

```python
dfburbal = pd.read_csv('./input/bureau_balance.csv')
dfburbal
```

# Get max bureau Balance 

```python
dfburmax = dfburbal.loc[dfburbal.groupby('SK_ID_BUREAU')['MONTHS_BALANCE'].idxmin()]
dfburmax = dfburmax.reset_index(drop=True)
dfburmax
```

## merge bureau balance and bureau

```python
dfbureau = pd.merge(dfbur, dfburmax, on='SK_ID_BUREAU', how='inner')
dfbureau
```

```python
# export merged Bureau to csv
dfbureau.to_csv('./output/bureau.csv', index=False)
```

```python
dfcreditcard = pd.read_csv('./input/credit_card_balance.csv')
dfcreditcard
```

# aggregate credit DF to save space

```python
dfaggcredit = dfcreditcard.groupby(['SK_ID_PREV', 'SK_ID_CURR']).agg({
    'AMT_BALANCE': 'sum',  # Calculate the mean of 'value1'
    'AMT_CREDIT_LIMIT_ACTUAL': 'sum',
    'AMT_DRAWINGS_ATM_CURRENT': 'sum',# Calculate the sum of 'value2'
    'AMT_DRAWINGS_CURRENT': 'sum',
    'AMT_DRAWINGS_POS_CURRENT': 'sum',
    'AMT_INST_MIN_REGULARITY': 'sum',
    'AMT_PAYMENT_CURRENT': 'sum',
    'AMT_PAYMENT_TOTAL_CURRENT':'sum',
    'AMT_RECEIVABLE_PRINCIPAL':'sum',
    'AMT_RECIVABLE': 'sum',
    'AMT_TOTAL_RECEIVABLE': 'sum',
    'CNT_DRAWINGS_ATM_CURRENT': 'sum',
    'CNT_DRAWINGS_CURRENT': 'sum',
    'CNT_DRAWINGS_OTHER_CURRENT': 'sum',
    'CNT_DRAWINGS_POS_CURRENT': 'sum',
    'CNT_INSTALMENT_MATURE_CUM': 'sum',
    'SK_DPD':'sum',
    'SK_DPD_DEF':'sum'
}).reset_index()

dfaggcredit
```

```python
#export aggregated credit card DF to csv
dfaggcredit.to_csv('./output/creditcard.csv', index=False)
```

# aggregate installment payments DF to save space

```python
dfinstall = pd.read_csv('./input/installments_payments.csv')
dfinstall
```

```python
dfagginstall = dfinstall.groupby(['SK_ID_PREV','SK_ID_CURR']).agg({
    'AMT_INSTALMENT':'sum',
    'AMT_PAYMENT':'sum'
}).reset_index()

dfagginstall['PAID_OFF'] = np.where(dfagginstall['AMT_PAYMENT'] >= dfagginstall['AMT_INSTALMENT'], 1, 0)
dfagginstall
```

```python
# expoert installments aggregate DF to csv
dfagginstall.to_csv('./output/installments.csv', index=False)
```

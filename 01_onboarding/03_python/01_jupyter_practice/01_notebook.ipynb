### Functions

```python
def read_my_csv_file(str_local_path):
    print(f'Reading file from {str_local_path}...')
    df = pd.read_csv(str_local_path)
    return df
```

```python
def tag_franchise(str_dealer_type):
    if str_dealer_type == 'Franchise':
        return 1
    else:
        return 0
```

```python
def tag_dealer_type(str_dealer_type, str_target_dealer_type):
    if str_dealer_type == str_target_dealer_type:
        return 1
    else:
        return 0
```

```python
class PlotProportions:
    def __init__(self):
        pass
    def create_proportion_table(self, df):
        # get nrows in df
        int_nrows = df.shape[0]
        # save to self
        self.int_nrows = int_nrows
        
        # get series
        ser_freq = df['DealerType'].value_counts(normalize=True)
        # convert series to df
        df_freq = ser_freq.reset_index()
        # rename the columns
        df_freq.columns = ['dealer_type', 'prop']
        # save the object to self
        self.df_freq = df_freq
        
        # return object
        return self
    def plot_proportions(self, str_local_path, bool_show_plot=True):
        # create
        fig, ax = plt.subplots(figsize=(9,5))
        # title
        ax.set_title(f'Proportion of Dealer Types (N = {self.int_nrows})')
        # x
        ax.set_xlabel('Dealer Type')
        # y
        ax.set_ylabel('Proportion')
        # plot
        bars = ax.bar(self.df_freq['dealer_type'], self.df_freq['prop'])
        for bar in bars:
            height = bar.get_height()
            ax.annotate(f'{height:0.2f}',
                xy=(bar.get_x() + bar.get_width() / 2, height),
                xytext=(0, 3),  # 3 points vertical offset
                textcoords="offset points",
                ha='center',
                va='bottom',
            )

        # save to self
        self.fig = fig
        
        # save
        plt.savefig(
            str_local_path, 
            bbox_inches='tight',
        )
        
        # show
        if bool_show_plot:
            plt.show()
        else:
            pass
```

```python
class Student(object):
    #"initialize the attributes for an object"
    def __init__(self, name, age, gender, level, grades=None):
        self.name = name
        self.age = age
        self.gender = gender
        self.level = level
        self.grades = grades or {}
    # add grades to a dictionary
    def setGrade(self, course, grade):
        self.grades[course] = grade
    # call a course and output grade    
    def getGrade(self, course):
        return self.grades[course]
    # calculate gpa
    def getGPA(self):
        return sum(self.grades.values())/len(self.grades)
    
# Define students & grades
john = Student("John", 12, "male", 6, {"math":3.3, "science":3.7})
jane = Student("Jane", 12, "female", 6, {"math":3.5, "science":4.0})
# Use new methods
print(jane.getGPA())
print(john.getGrade("math"))
john.setGrade("history", 4.0)
print(john.getGPA())
```

### Tag franchise for example of applying function row-wise (one arg)

```python
df['tag_franchise'] = df.apply(
    lambda x: tag_franchise(
        str_dealer_type=x['DealerType'],
    ),
    axis=1,
)
df
```

### Tag independent for example of applying function row-wise (multiple args)

```python
df['tag_independent'] = df.apply(
    lambda x: tag_dealer_type(
        str_dealer_type=x['DealerType'],
        str_target_dealer_type='Independent',
    ),
    axis=1,
)
df
```

### Make multiple tags in a dynamic list

```python
# create a list of the unique values in column
list_str_dealer_type = list(df['DealerType'].value_counts().index)

# create a tag column for each unique value in column
for str_dealer_type in tqdm(list_str_dealer_type):
    df[f'tag_{str_dealer_type}'] = df.apply(
        lambda x: tag_dealer_type(
            str_dealer_type=x['DealerType'],
            str_target_dealer_type=str_dealer_type,
        ),
        axis=1,
    )
# show
df
```

### Iterate and create a df

```python
# # get list of cols
# list_cols = list(df.columns)
# for a, col in enumerate(list_cols):
#     print(f'{a+1} - {col}')
```

```python
# get random sample of rows
df_tmp = df.sample(frac=0.01, random_state=42)

# get cols
list_cols = list(df_tmp.columns)

# iterate and create df
list_dict_row = []
for col in tqdm(list_cols):
    try:
        # get mean
        flt_mean = df_tmp[col].mean()
        # get median
        flt_mdn = df_tmp[col].median()
        # create dict
        dict_row = {
            'col': col,
            'mean': flt_mean,
            'median': flt_mdn,
        }
        list_dict_row.append(dict_row)
    except TypeError:
        pass

# make df
df_tmp = pd.DataFrame(list_dict_row)
# show
df_tmp
```
